```python
import numpy as np

def generate_graphene_coordinates(num_rows, num_cols, size):
    """Generates the coordinates of atoms for a graphene flake."""
    coordinates = []
    for i in range(num_rows):
        for j in range(num_cols):
            x = j * size * 3**0.5
            y = i * size * 3 / 2 * 1.5  # Increase 1.5 to adjust spacing between rows
            if i % 2 == 0:
                coordinates.append([x, y, 0])
            else:
                coordinates.append([x + size * 3**0.5 / 2, y, 0])
    return np.array(coordinates)

def save_xyz_file(coordinates, file_name):
    """Saves the coordinates to an XYZ file."""
    with open(file_name, 'w') as f:
        f.write(f"{len(coordinates)}\n")
        f.write("Generated graphene flake\n")
        for coord in coordinates:
            f.write(f"C {coord[0]} {coord[1]} {coord[2]}\n")

# Number of rows and columns in the graphene flake
num_rows = 5
num_cols = 5
# Size of each hexagon
hexagon_size = 1.0

# Generate coordinates
graphene_coordinates = generate_graphene_coordinates(num_rows, num_cols, hexagon_size)

# Save coordinates to XYZ file
save_xyz_file(graphene_coordinates, 'graphene_flake.xyz')
```

```python
def generate_graphene_coordinates(num_rows, num_cols, bond_length):
    """Generates the coordinates of atoms for a graphene flake."""
    coordinates = []
    for i in range(num_rows):
        for j in range(num_cols):
            x = j * bond_length * 3**0.5
            y = i * bond_length * 3 / 2 * 1.5  # Increase 1.5 to adjust spacing between rows
            if i % 2 == 0:
                coordinates.append([x, y, 0])
            else:
                coordinates.append([x + bond_length * 3**0.5 / 2, y, 0])
    return np.array(coordinates)

# Number of rows and columns in the graphene flake
num_rows = 5
num_cols = 5
# Bond length in Ångström
bond_length = 1.42

# Generate coordinates
graphene_coordinates = generate_graphene_coordinates(num_rows, num_cols, bond_length)

# Save coordinates to XYZ file
save_xyz_file(graphene_coordinates, 'graphene_flake.xyz')
```

```python
import numpy as np
import matplotlib.pyplot as plt

def generate_graphene_flake(num_rows, num_columns, bond_length):
    # Calcola le coordinate degli atomi di carbonio
    coordinates = []
    for i in range(num_rows):
        for j in range(num_columns):
            x = j * np.sqrt(3) * bond_length
            y = i * 1.5 * bond_length
            if i % 2 == 1:
                x += np.sqrt(3) / 2 * bond_length
            coordinates.append((x, y))

    return coordinates

def plot_graphene_flake(coordinates, bond_length):
    fig, ax = plt.subplots()
    ax.set_aspect('equal')
    ax.set_axis_off()

    # Disegna gli atomi di carbonio
    for x, y in coordinates:
        hexagon = plt.RegularPolygon((x, y), numVertices=6, radius=bond_length, edgecolor='black', facecolor='none')
        ax.add_patch(hexagon)

    plt.show()

# Definisci le dimensioni del flake di grafene
num_rows = 10
num_columns = 10
bond_length = 1.42  # Raggio di covalenza del carbonio in angstrom

# Genera il flake di grafene
coordinates = generate_graphene_flake(num_rows, num_columns, bond_length)
def write_xyz_file(coordinates, filename):
    num_atoms = len(coordinates)
    with open(filename, 'w') as f:
        f.write(f"{num_atoms}\n")
        f.write("Generated graphene flake\n")
        for x, y in coordinates:
            f.write(f"C {x} {y} 0.0\n")

# Genera il flake di grafene

# Scrivi il file XYZ
write_xyz_file(coordinates, "graphene_flake.xyz")
```

```python
def get_xyz_types(file_path:Path)->"list[str]":
    atom_types = []

    with open(file_path, "r") as f:
        lines = f.readlines()

    num_atoms = int(lines[0])

    for line in lines[2:2 + num_atoms]:
        atom_symbol = line.split()[0]
        if atom_symbol not in atom_types:
            atom_types.append(atom_symbol)

    return atom_types
    
    
get_xyz_types(Path("graphene_flake.xyz"))
    
```
